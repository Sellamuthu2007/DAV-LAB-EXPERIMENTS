```python
import pandas as pd
# Read data
text_df = pd.read_csv('/content/sample_data/california_housing_train.csv')

excel_df = pd.read_csv('/content/sample_data/mnist_train_small.csv') # Using a CSV as a placeholder for the missing excel file
web_df = pd.read_csv('https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv')
# Display data
print(text_df.head(), "\n", excel_df.head(), "\n", web_df.head())
# Handle missing values
text_df.ffill(inplace=True);
excel_df.bfill(inplace=True);
web_df.dropna(inplace=True);
# Save processed data
text_df.to_csv('processed_text.csv', index=False)
# excel_df.to_excel('processed_excel.xlsx', index=False) # Commenting out as we are using a CSV for excel_df now
```
